```python
# import library
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
```

## Langkah 2 - Membuat Data Dummy

```python
from sklearn.datasets import make_blobs
X, y = make_blobs(n_samples=50, centers=2,
                  random_state=0, cluster_std=0.60)
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
```

## Langkah 4 - Ilustrasi Margin

```python
xfit = np.linspace(-1, 3.5)
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')

for m, b, d in [(1, 0.65, 0.33), (0.5, 1.6, 0.55), (-0.2, 2.9, 0.2)]:
    yfit = m * xfit + b
    plt.plot(xfit, yfit, '-k')
    plt.fill_between(xfit, yfit - d, yfit + d, edgecolor='none',
                     color='#AAAAAA', alpha=0.4)

plt.xlim(-1, 3.5)
```

## Langkah 5 - Fitting Model

```python
from sklearn.svm import SVC # "Support vector classifier"
model = SVC(kernel='linear', C=1E10)
model.fit(X, y)
```

```python
# buat sebuah fungsi untuk menampilkan fitting data

def plot_svc_decision_function(model, ax=None, plot_support=True):
    
    if ax is None:
        ax = plt.gca()
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()
    
    # buat grid untuk evaluasi model
    x = np.linspace(xlim[0], xlim[1], 30)
    y = np.linspace(ylim[0], ylim[1], 30)
    Y, X = np.meshgrid(y, x)
    xy = np.vstack([X.ravel(), Y.ravel()]).T
    P = model.decision_function(xy).reshape(X.shape)
    
    # plot batas dan margin
    ax.contour(X, Y, P, colors='k',
               levels=[-1, 0, 1], alpha=0.5,
               linestyles=['--', '-', '--'])
    
    # plot support vectors
    if plot_support:
        ax.scatter(model.support_vectors_[:, 0],
                   model.support_vectors_[:, 1],
                   s=300, linewidth=1, facecolors='none');
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
plot_svc_decision_function(model)
```

```python
model.support_vectors_
```

```python
def plot_svm(N=10, ax=None):
    X, y = make_blobs(n_samples=200, centers=2,
                      random_state=0, cluster_std=0.60)
    X = X[:N]
    y = y[:N]
    model = SVC(kernel='linear', C=1E10)
    model.fit(X, y)
    
    ax = ax or plt.gca()
    ax.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
    ax.set_xlim(-1, 4)
    ax.set_ylim(-1, 6)
    plot_svc_decision_function(model, ax)

fig, ax = plt.subplots(1, 2, figsize=(16, 6))
fig.subplots_adjust(left=0.0625, right=0.95, wspace=0.1)
for axi, N in zip(ax, [60, 120]):
    plot_svm(N, axi)
    axi.set_title('N = {0}'.format(N))
```

```python
# jumlah data dapat dipilih di antara 10 atau 200 buah data, telihat tidak ada perubahan pada model

from ipywidgets import interact, fixed
interact(plot_svm, N=[10, 200], ax=fixed(None))
```

## Evaluasi dan analisis tambahan

```python
print('Support vectors per kelas:', model.n_support_)
print('Akurasi training:', model.score(X, y))
```

Titik pada batas margin menjadi support vectors. Penambahan titik yang jauh dari margin dapat mempertahankan pemisah; support vector baru dapat mengubahnya. Seluruh skor pada contoh ini memakai data training sintetis.

## Dataset pilihan sendiri - Diagnosis kanker payudara Wisconsin

Pengantar modul meminta mahasiswa memilih dataset sendiri selain contoh dummy. Saya memilih **Breast Cancer Wisconsin (Diagnostic)** dari [UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/17/breast+cancer+wisconsin+diagnostic), tersedia lokal melalui `load_breast_cancer`. Dataset memuat 569 sampel dengan 30 fitur numerik dan dua kelas diagnosis (malignant/benign), sehingga sesuai untuk klasifikasi SVM. Semua fitur dipakai untuk model; dua fitur divisualisasikan untuk inspeksi awal.

Data dibagi stratified 70:30 dengan seed 42. StandardScaler dipelajari hanya pada training melalui pipeline. Kernel linier dipilih mengikuti fokus Praktikum 1. Hasil ini merupakan latihan klasifikasi pada dataset publik.


```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay

chosen_data = load_breast_cancer()
print('Sampel dan fitur:', chosen_data.data.shape)
print('Kelas:', chosen_data.target_names)
plt.figure()
plt.scatter(chosen_data.data[:, 0], chosen_data.data[:, 1], c=chosen_data.target, cmap='coolwarm', alpha=0.6)
plt.xlabel(chosen_data.feature_names[0])
plt.ylabel(chosen_data.feature_names[1])
plt.title('Inspeksi dua fitur dataset pilihan')
plt.show()
chosen_X_train, chosen_X_test, chosen_y_train, chosen_y_test = train_test_split(
    chosen_data.data, chosen_data.target, test_size=0.3, stratify=chosen_data.target, random_state=42)
chosen_model = make_pipeline(StandardScaler(), SVC(kernel='linear', C=1))
chosen_model.fit(chosen_X_train, chosen_y_train)
chosen_pred = chosen_model.predict(chosen_X_test)
print('Akurasi train:', chosen_model.score(chosen_X_train, chosen_y_train))
print('Akurasi test:', accuracy_score(chosen_y_test, chosen_pred))
print(classification_report(chosen_y_test, chosen_pred, target_names=chosen_data.target_names))
ConfusionMatrixDisplay.from_predictions(chosen_y_test, chosen_pred, display_labels=chosen_data.target_names)
plt.show()
print('Support vectors per kelas:', chosen_model.named_steps['svc'].n_support_)
```

### Analisis dataset pilihan

SVM linier menggunakan seluruh 30 fitur yang telah distandardisasi untuk memisahkan diagnosis. Visualisasi dua fitur hanya menunjukkan sebagian informasi dan bukan batas keputusan model 30 dimensi. Confusion matrix memperlihatkan kesalahan pada setiap kelas; recall malignant perlu diperhatikan selain akurasi. Evaluasi berlaku pada split ini dan tidak membuktikan kelayakan pemakaian klinis.

## Langkah 5: Visualisasi Data

```python
# import library untuk visualisasi
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# visualisasi data dengan pairplot
sns.pairplot(data, x_vars=['Time on App', 'Time on Website', 'Length of Membership'],
             y_vars='Yearly Amount Spent', height=4, aspect=1, kind='scatter')
plt.show()
```

## Langkah 6: Regresi Linier

```python
# Membuat variabel bebas X dan Y, contoh pengambilan dari analisis korelasi sebelumnya
X = data['Length of Membership']
y = data['Yearly Amount Spent']
```

```python
# Pembagian data latih dan data uji dengan proporsi 7:3
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, test_size=0.3, random_state=100)
```

```python
# Training model
import statsmodels.api as sm

X_train_sm = sm.add_constant(X_train)
lr = sm.OLS(y_train, X_train_sm).fit()
```

```python
# Visualisasi garis regresi
plt.scatter(X_train, y_train)
plt.plot(X_train, lr.predict(X_train_sm), 'r')
plt.show()
```

## Langkah 7: Analisis Residual

```python
# Prediksi nilai y_value dari data x yang telah dilatih
y_train_pred = lr.predict(X_train_sm)

# Menghitung residual
res = (y_train - y_train_pred)
```

```python
# Histogram residual
fig = plt.figure()
sns.histplot(res, bins=15, kde=True)
plt.title('Error Terms', fontsize=15)
plt.xlabel('y_train - y_train_pred', fontsize=15)
plt.show()

# Scatter plot residual
plt.scatter(X_train, res)
plt.show()
```

## Langkah 8: Prediksi pada Data Uji dan Evaluasi Model

```python
# Prediksi pada data uji
X_test_sm = sm.add_constant(X_test)
y_test_pred = lr.predict(X_test_sm)
```

```python
# Hitung nilai R-squared
from sklearn.metrics import r2_score

r_squared = r2_score(y_test, y_test_pred)
```

## Langkah 9: Visualisasi Hasil

```python
# Visualisasi data uji dan hasil prediksi
plt.scatter(X_test, y_test)
plt.plot(X_test, y_test_pred, 'r')
plt.show()
```

## Evaluasi dan analisis tambahan

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
from IPython.display import display, Markdown
print(lr.summary())
print('R-squared test:', r_squared)
display(pd.DataFrame({'MAE': [mean_absolute_error(y_test, y_test_pred)],
                      'MSE': [mean_squared_error(y_test, y_test_pred)],
                      'RMSE': [np.sqrt(mean_squared_error(y_test, y_test_pred))],
                      'R-squared': [r_squared]}, index=['Test 30%']))
display(Markdown(f'Model memperoleh **R² uji {r_squared:.4f}**. Koefisien masa keanggotaan '
                 f'**{lr.params.iloc[1]:.2f}** menunjukkan asosiasi positif dengan pengeluaran. '
                 'Residual tetap memiliki variasi karena fitur perilaku lain belum dimasukkan.'))
```
